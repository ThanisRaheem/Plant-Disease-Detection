# Data preprocessing ? Custom CNN
Reproducible 70/15/15 stratified split; source files are never modified.

```python
from pathlib import Path
import csv, random
from collections import Counter
from sklearn.model_selection import train_test_split
RANDOM_SEED=42; ROOT=Path.cwd().parent if Path.cwd().name=="notebooks" else Path.cwd(); DATASET=ROOT/"PlantVillage"; OUT=ROOT/"results/custom_cnn"; OUT.mkdir(parents=True,exist_ok=True)
EXTS={".jpg",".jpeg",".png",".bmp",".gif",".tif",".tiff"}; rows=[]
for d in sorted(p for p in DATASET.iterdir() if p.is_dir()):
 for p in sorted(d.rglob("*")):
  if p.is_file() and p.suffix.lower() in EXTS: rows.append((str(p.resolve()),d.name))
classes=sorted({x[1] for x in rows}); ci={c:i for i,c in enumerate(classes)}
paths=[x[0] for x in rows]; labels=[ci[x[1]] for x in rows]
tr, rem=train_test_split(range(len(rows)),test_size=.30,stratify=labels,random_state=RANDOM_SEED); va,te=train_test_split(rem,test_size=.50,stratify=[labels[i] for i in rem],random_state=RANDOM_SEED)
split={i:"train" for i in tr}|{i:"validation" for i in va}|{i:"test" for i in te}; allrows=[[paths[i],rows[i][1],labels[i],split[i]] for i in range(len(rows))]
def w(name,data):
 with open(OUT/name,"w",newline="",encoding="utf-8") as f: csv.writer(f).writerows([["filepath","class_name","class_index","split"],*data])
w("dataset_split.csv",allrows); w("train_metadata.csv",[r for r in allrows if r[3]=="train"]); w("validation_metadata.csv",[r for r in allrows if r[3]=="validation"]); w("test_metadata.csv",[r for r in allrows if r[3]=="test"])
print(len(classes),Counter(r[3] for r in allrows),"overlap",len(set(paths[ i] for i in tr)&set(paths[i] for i in va)))
```
